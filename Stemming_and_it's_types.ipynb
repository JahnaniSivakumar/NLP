```python
words = ['eating', 'eat', 'eaten','history', 'finally', 'finalize', 'going', 'gone', 'writing', 'writes', 'programming', 'programs']
```

## PorterStemmer

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
```

```python
for w in words:
    print(w + ' : ' + ps.stem(w))
```

For some words, we cannot get the exact stem words.

```python
ps.stem('Congratulations')
```

No meaning here for the word

## RegexpStemmer

```python
from nltk.stem import RegexpStemmer
rs = RegexpStemmer('ing$|s$|e$|able$', min = 4)
```

```python
rs.stem('coughing')
```

```python
rs.stem('ingplaying')
```

## SnowballStemmer

```python
from nltk.stem import SnowballStemmer
ss = SnowballStemmer('english')
```

```python
for w in words:
    print(w + ' : ' + ss.stem(w))
```

## Porter Vs Snowball

```python
ps.stem('fairly'), ps.stem('sportingly'), ps.stem('lying'), ps.stem('goes')
```

```python
ss.stem('fairly') , ss.stem('sportingly') , ss.stem('lying'), ss.stem('goes')
```

Some of the words in Snowball Stemmer also faces erros in finding the right stem word
